# Decklist to SPELLS

Run all cells to query Scryfall and print a copyable `SPELLS` tuple for the mana evaluator. Replace `DECKLIST` below, then rerun the input and import cells.

Only main-deck **nonlands** are extracted. Exact card/front-face names are used; set codes, collector numbers, and format legality are ignored. Lands of every family are excluded, including utility lands. An optional `Deck` heading is accepted; the `Sideboard` section is excluded.

Printed costs must contain only generic mana and WUBRG pips. For transforming cards, modal double-faced cards, adventures, and flip cards, only the front/main face is considered. Alternate costs and back-face costs are ignored; split cards and unsupported cost symbols are rejected.

Every import queries Scryfall afresh, requires internet access, and saves no card catalog or cache. A failed lookup or unsupported nonland prints diagnostics and leaves no `SPELLS` value. Successful entries have weight 1; duplicate cards combine, while different cards with equal costs remain separate.

The supplied example has 60 main-deck cards: 37 nonlands across 15 names and 23 lands. Actual costs always come from the live API. [Scryfall API documentation](https://scryfall.com/docs/api).

```
from mtg_tools.decklist_import import DecklistImportError, format_spells, import_decklist
from mtg_tools.mana_opportunities import SpellCost
```

```
DECKLIST = '''4 Kaito, Bane of Nightmares (DSK) 220
4 Spyglass Siren (LCI) 78
2 Enduring Curiosity (DSK) 51
2 Great Gilded Boat (HOB) 42
2 Bitter Triumph (SOA) 26
2 Shoot the Sheriff (OTJ) 106
2 Tishana's Tidebinder (LCI) 81
1 Spell Pierce (XLN) 81
4 Desolation Prowler (HOB) 64
2 Cecil, Dark Knight (FIN) 91
3 Requiting Hex (ECL) 116
2 Dream Beavers (TMT) 62
4 Floodpits Drowner (DSK) 59
1 Spell Snare (ECL) 71
2 The Wondrous Wasp (MSH) 84
4 Watery Grave (GRN) 259
4 Gloomlake Verge (DSK) 260
2 Soulstone Sanctuary (FDN) 133
2 Restless Reef (LCI) 282
4 Island (ZNR) 381
4 Swamp (ZNR) 382
1 Fountainport (BLB) 253
1 Hidden Lair (MSH) 269
1 Shipwreck Marsh (MID) 267'''
```

```
# Remove earlier results before querying, so failures cannot leave stale spells.
globals().pop('SPELLS', None)
globals().pop('imported', None)

try:
    imported = import_decklist(DECKLIST)
except DecklistImportError as error:
    print(error)
else:
    SPELLS = imported.spells
    print(format_spells(imported))
    print()
    nonland_count = sum(spell.count for spell in SPELLS)
    print(f'{nonland_count} nonlands across {len(SPELLS)} names; {imported.excluded_land_count} lands excluded.')
    print(f'Main-deck total: {nonland_count + imported.excluded_land_count}')
```
